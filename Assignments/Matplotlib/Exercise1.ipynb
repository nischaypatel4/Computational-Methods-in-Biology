## Exercise 1

**Follow along with these steps**
* Create a figure object called fig using plt.figure()
* Use add_axes to add an axis to the figure canvas at [0,0,1,1]. Call this new axis ax.
* Plot (x,y) on that axes and set the labels and titles to match the plot below:

```python
import matplotlib.pyplot as plt
fig = plt.figure()
ax = fig.add_axes([0, 0, 1, 1]) # left, bottom, width, height (range 0 to 1) and fig.add_axes()is used to add axes and plot
ax.plot(x, y, 'b')#axes.plot() is used to plot the graph
ax.set_xlabel('X') # Notice the use of set_ to begin methods
ax.set_ylabel('y')
ax.set_title('Set Title')
```

## Exercise 4

**Use plt.subplots(nrows=1, ncols=2) to create the plot below.**

```python
# Use similar to plt.figure() except use tuple unpacking to grab fig and axes
fig, axes = plt.subplots() #returns a tuple containing a Figure object and an array of Axes objects

# Now use the axes object to add stuff to plot
axes.plot(x, y, 'r')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('title')
```

**Now plot (x,y) and (x,z) on the axes. Play around with the linewidth and style**

```python
fig, axes = plt.subplots(1,2)
x=np.linspace(0, 101, 25)
y=2*x
axes[0].plot(x, y,'b--')
axes[0].set_xlabel('x')
axes[0].set_ylabel('z')
axes[0].set_xlim([0, 100])  # Set x-axis limits
axes[0].set_ylim([0, 200])
y=x**2
axes[1].plot(x,y , 'r')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].set_xlim([0, 100])  # Set x-axis limits
axes[1].set_ylim([0, 10000])
```

**See if you can resize the plot by adding the figsize() argument in plt.subplots() are copying and pasting your previous code.**

```python
fig,axes=plt.subplots(nrows=1,ncols=2,figsize=(8,2))
x=np.linspace(0, 101, 25)
y=2*x
axes[0].plot(x, y,'b')

axes[0].set_xlabel('x')
axes[0].set_ylabel('z')
axes[0].set_xlim([0, 100])  # Set x-axis limits
axes[0].set_ylim([0, 200])
y=x**2
axes[1].plot(x,y , 'r--')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].set_xlim([0, 100])  # Set x-axis limits
axes[1].set_ylim([0, 10000])
```
